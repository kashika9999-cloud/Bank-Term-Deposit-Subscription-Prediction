```python
df['previously_contacted'] = (df['pdays'] != 999).astype(int)

df_model = df.drop(columns=['pdays', 'duration'])

print(df_model.shape)
print(df_model.columns)
```

```python
from sklearn.model_selection import train_test_split

X = df_model.drop(columns=['y']) # predictors
y = df_model['y'] # predicted value

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:")
print(y_train.value_counts(normalize=True))
print("y_test:")
print(y_test.value_counts(normalize=True))
```

```python
categorical_cols = X_train.select_dtypes(include='object').columns
numerical_cols = X_train.select_dtypes(exclude='object').columns

print("Categorical:", list(categorical_cols))
print("Numerical:", list(numerical_cols))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

log_reg = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('classifier', LogisticRegression(max_iter=1000))
    ]
)
```

```python
log_reg.fit(X_train, y_train)
```

```python
y_pred = log_reg.predict(X_test)
```

```python
y_prob = log_reg.predict_proba(X_test)[:, 1]
y_prob
```

```python
feature_names = log_reg.steps[0][1].get_feature_names_out()
coefficients = log_reg.steps[-1][1].coef_[0]

coef_df = pd.DataFrame({
    'Feature': feature_names,
    'Coefficient': coefficients
})

coef_df
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.metrics import make_scorer, f1_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline

# K values to test
k_values = [3, 5, 7, 9, 11, 15, 21]

# F1 scorer for the "yes" class
f1_yes = make_scorer(f1_score, pos_label='yes')

cv_scores = []

for k in k_values:

    knn_cv = Pipeline([
        ('preprocessor', preprocessor),
        ('knn', KNeighborsClassifier(n_neighbors=k))
    ])

    scores = cross_val_score(
        knn_cv,
        X_train,
        y_train,
        cv=5,
        scoring=f1_yes
    )

    cv_scores.append(scores.mean())

for k, score in zip(k_values, cv_scores):
    print("K =", k, "Mean CV F1 for 'yes' =", score) # K=11 IS THE BEST BUT STILL LESS EFFICIENT
```

```python
# TUNING THE HYPERPARAMETERS
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline

dt_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('tree', DecisionTreeClassifier(random_state=42))
])

param_grid = {
    'tree__max_depth': [3, 5, 7, 10, 15, None],
    'tree__min_samples_split': [2, 5, 10, 20],
    'tree__min_samples_leaf': [1, 2, 5, 10, 20],
    'tree__criterion': ['gini', 'entropy']
}
```

```python
from sklearn.metrics import make_scorer, f1_score

f1_yes = make_scorer(f1_score, pos_label='yes')

grid_search = GridSearchCV(
    dt_pipeline,
    param_grid,
    cv=5,
    scoring=f1_yes,
    n_jobs=-1
)

grid_search.fit(X_train, y_train)
```

```python
print("Best parameters:", grid_search.best_params_)
print("Best CV F1:", grid_search.best_score_)
```

```python
best_dt = grid_search.best_estimator_

y_pred_dt_tuned = best_dt.predict(X_test)
y_prob_dt_tuned = best_dt.predict_proba(X_test)[:, 1]
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_dt_tuned))
print("Precision:", precision_score(y_test, y_pred_dt_tuned, pos_label='yes'))
print("Recall:", recall_score(y_test, y_pred_dt_tuned, pos_label='yes'))
print("F1 Score:", f1_score(y_test, y_pred_dt_tuned, pos_label='yes'))
print("ROC-AUC:", roc_auc_score(y_test, y_prob_dt_tuned))
```
